```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['name'] = df['name'].str.strip()
```

```python
df['planet_status'] = df['planet_status'].str.strip()
```

```python
df['planet_status'].value_counts()
```

```python
df.dtypes
```

```python
df['mass'] = pd.to_numeric(df['mass'], errors='coerce')
```

```python
df['radius'] = pd.to_numeric(df['radius'], errors='coerce')
```

```python
numeric_columns = [
    'mass_error_min',
    'mass_error_max',
    'mass_sini',
    'mass_sini_error_min',
    'mass_sini_error_max',
    'radius_error_min'
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.dropna(subset=['name', 'planet_status'], inplace=True)
```
